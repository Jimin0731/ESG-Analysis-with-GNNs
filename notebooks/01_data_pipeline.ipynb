# Data pipeline and typed contracts

`SupervisedSnapshot` and `PreparedSupervisedData` preserve chronological split, node, feature, target, directed-edge, economic-weight, and observed-label contracts. Upstream loaders, graph snapshots, feature panels, target panels, train-only preprocessing, and direct-lineage leakage audit produce these contracts; they are not reconstructed here.

```python
from examples.notebook_support import build_synthetic_prepared_data
from src.experiments import SupervisedSnapshot, PreparedSupervisedData
data = build_synthetic_prepared_data()
assert isinstance(data, PreparedSupervisedData) and isinstance(data.train[0], SupervisedSnapshot)
```

```python
rows=[]
for split in ('train','validation','test'):
    for s in getattr(data, split):
        rows.append({'period':s.period,'split':split,'node count':len(s.node_ids),'feature width':s.x.shape[1],'target width':s.y.shape[1],'observed target counts':tuple(int(s.target_observed_mask[:,i].sum()) for i in range(s.y.shape[1])),'edge count':s.edge_index.shape[1]})
rows
```

```python
s=data.train[1]
print('period order:', tuple(x.period for x in data.train + data.validation + data.test))
print('node order:', s.node_ids)
print('feature order:', s.feature_names)
print('target order:', s.target_names)
print('observed mask:', s.target_observed_mask)
print('source -> target:', tuple(zip(s.edge_index[0].tolist(),s.edge_index[1].tolist())))
print('aligned economic weights:', s.edge_weight.tolist())
```

Missing targets remain missing and are represented by the observed-target mask; they are never imputed. Train, validation, and test snapshots remain separate and exactly ordered.

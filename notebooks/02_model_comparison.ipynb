# Small model comparison

This fixed smoke list is not hyperparameter optimization. MLP is a non-graph neural baseline; train mean is non-neural; weighted GAT uses economic edge weights; GPR propagation coefficients are not attention. Tiny synthetic results are smoke validation, not research findings.

```python
from examples.notebook_support import build_synthetic_prepared_data, build_example_model_config, run_example_training, run_example_baseline
from src.models import available_models, build_model
data=build_synthetic_prepared_data()
selected=("mlp","weighted_gat","gpr_gnn")
print("registry reference:", available_models())
for name in selected: build_model(build_example_model_config(name))
```

```python
rows=[]
for name in selected:
    result=run_example_training(name,epochs=2,prepared_data=data)
    val=result.evaluations['validation']; test=result.evaluations['test']
    rows.append({'model':name,'validation MAE':val.macro_mae,'validation RMSE':val.macro_rmse,'test MAE (descriptive only)':test.macro_mae,'test RMSE (descriptive only)':test.macro_rmse})
b=run_example_baseline(data)
rows.append({'model':'train_mean','validation MAE':b.evaluations['validation'].macro_mae,'validation RMSE':b.evaluations['validation'].macro_rmse,'test MAE (descriptive only)':b.evaluations['test'].macro_mae,'test RMSE (descriptive only)':b.evaluations['test'].macro_rmse})
rows
```

Every run preserves identical feature and target order and uses the same regression metrics. Validation results can check workflow behavior; test values are final descriptive reporting only. No model is labeled scientifically superior.

# Directed interpretability

> **Model association/local sensitivity — not a causal estimate**

This validation-only example keeps `prediction_delta`, `aggregated_attention_coefficient`, `attention_path_score`, and `economic_edge_weight` distinct. Processed feature changes may not correspond to one natural economic unit.

```python
import matplotlib.pyplot as plt
from examples.notebook_support import build_synthetic_prepared_data, run_example_training, reconstruct_trained_model, build_example_explanation
from src.evaluation import build_shock_explanation
from src.visualization import plot_shock_prediction_changes
data=build_synthetic_prepared_data()
result=run_example_training("weighted_gat",epochs=3,prepared_data=data)
model=reconstruct_trained_model(result)
snapshot=data.validation[0]
bundle=build_example_explanation(model,snapshot)
```

```python
print('normalization:',bundle.attention.normalization,'valid:',bundle.attention.normalization_valid)
print('directed final-layer attention:',[(e.source_node_id,e.target_node_id,e.aggregated_attention_coefficient,e.economic_edge_weight) for e in bundle.attention.edges])
print('prediction deltas:',[(r.node_id,r.target_name,r.prediction_delta) for r in bundle.perturbation.records])
print('bounded paths:',[(r.path_node_ids,r.attention_path_score) for r in bundle.attention_paths.records])
print('feature ablation:',bundle.feature_ablation.records)
print('edge-weight ablation:',bundle.edge_ablation.records)
```

```python
figure=plot_shock_prediction_changes(bundle.perturbation,target_name=data.target_names[0],top_k=3)
figure
```

Attention is normalized per incoming target and head before head aggregation. The additive source-node perturbation, one feature ablation, and one edge-weight ablation are frozen-model local sensitivities. Bounded path products summarize attention routes; none of these quantities is a causal estimate. Nothing is exported by default.

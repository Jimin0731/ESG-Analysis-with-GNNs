# Chronological evaluation

The canonical trainer consumes ascending train periods, uses validation-only early stopping and scheduler decisions, keeps its best checkpoint in memory, optionally restores it, applies masked multi-target loss, globally aggregates observed-label counts, and handles undefined R² explicitly. Preprocessing was already fitted using train periods.

```python
from examples.notebook_support import build_synthetic_prepared_data, run_example_training
from src.experiments import train_supervised_model
data=build_synthetic_prepared_data()
print("ascending train periods:",tuple(s.period for s in data.train))
result=run_example_training("weighted_gat",epochs=4,prepared_data=data)
```

```python
history=[{'epoch':r.epoch,'train loss':r.train_total_loss,'validation loss':r.validation_total_loss,'learning rate':r.learning_rate,'became best':r.became_best} for r in result.history]
history
```

```python
for split in ('train','validation','test'):
 print(split,[(m.target_name,m.mae,m.rmse,m.r2,m.observed_count,m.r2_undefined_reason) for m in result.evaluations[split].metrics])
print('best epoch:',result.best_epoch,'restored best:',result.restored_best)
```

Test snapshots are not used during training, and test metrics do not select the best epoch. Missing targets are masked, not filled.
